```python
# =========================================
# SALES DATA STATISTICAL ANALYSIS PROJECT
# =========================================

# Import Libraries
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

# -----------------------------------------
# Sample Sales Dataset
# -----------------------------------------

sales = [120, 150, 180, 200, 220, 250, 270, 300, 320, 350]
advertisement = [10, 12, 15, 18, 20, 22, 25, 28, 30, 35]

# Create DataFrame
df = pd.DataFrame({
    "Sales": sales,
    "Advertisement": advertisement
})

print("DATASET")
print(df)

# =========================================
# 1. PERCENTILES
# =========================================

print("\n====================")
print("PERCENTILES")
print("====================")

p25 = np.percentile(df["Sales"], 25)
p50 = np.percentile(df["Sales"], 50)
p75 = np.percentile(df["Sales"], 75)

print("25th Percentile:", p25)
print("50th Percentile:", p50)
print("75th Percentile:", p75)

# =========================================
# 2. QUARTILES
# =========================================

print("\n====================")
print("QUARTILES")
print("====================")

quartiles = df["Sales"].quantile([0.25, 0.50, 0.75])

print(quartiles)

# =========================================
# 3. HISTOGRAM
# =========================================

print("\nGenerating Histogram...")

plt.hist(df["Sales"], bins=5)
plt.title("Sales Histogram")
plt.xlabel("Sales")
plt.ylabel("Frequency")
plt.show()

# =========================================
# 4. SKEWNESS
# =========================================

print("\n====================")
print("SKEWNESS")
print("====================")

skewness = df["Sales"].skew()

print("Sales Skewness:", skewness)

# =========================================
# 5. COVARIANCE
# =========================================

print("\n====================")
print("COVARIANCE")
print("====================")

covariance = np.cov(df["Sales"], df["Advertisement"])

print(covariance)

# =========================================
# 6. CORRELATION
# =========================================

print("\n====================")
print("CORRELATION")
print("====================")

correlation = df["Sales"].corr(df["Advertisement"])

print("Correlation:", correlation)

# =========================================
# 7. DEPENDENCY ANALYSIS
# =========================================

print("\n====================")
print("DEPENDENCY ANALYSIS")
print("====================")

if correlation > 0:
    print("Positive Dependency")
    print("Advertisement increases -> Sales increase")
else:
    print("Negative Dependency")

# =========================================
# 8. SCATTER PLOT
# =========================================

plt.scatter(df["Advertisement"], df["Sales"])

plt.title("Advertisement vs Sales")
plt.xlabel("Advertisement")
plt.ylabel("Sales")

plt.show()

print("\nPROJECT COMPLETED SUCCESSFULLY")
```
